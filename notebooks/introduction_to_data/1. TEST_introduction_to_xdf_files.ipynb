```python
selected_index = 0
selected_stream = streams[selected_index]

data = selected_stream["time_series"]
time = selected_stream["time_stamps"]
```

```python
time_insert = time - time[0]
```

```python
channels = selected_stream["info"]["desc"][0]["channels"][0]["channel"]
column_names = [ch["label"][0] for ch in channels]

df = pd.DataFrame(data, columns=column_names)
```

```python
df.insert(0, "time_stamp", time)
df.insert(1, "time_insert", time_insert)
```

```python
df.head()
```

```python
t=df["time_insert"]
plt.figure(figsize=(12,5))

plt.plot(t, df["ax"], label="ax")
plt.plot(t, df["ay"], label="ay")
plt.plot(t, df["az"], label="az")

plt.xlabel("time")
plt.ylabel("Acceleration")
plt.title("Accelerator data")

plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["time_insert"], df["ax"], label="ax")
plt.plot(df["time_insert"], df["ay"], label="ay")

plt.xlabel("time (s)")
plt.ylabel("horizontal acceleration")
plt.title("Horizontal acceleration: ax and ay")
plt.legend()
plt.grid(True)
plt.show()
```

```python
df["acc_horizontal"] = np.sqrt(df["ax"]**2 + df["ay"]**2)

plt.figure(figsize=(12, 5))
plt.plot(df["time_insert"], df["acc_horizontal"])

plt.xlabel("time (s)")
plt.ylabel("horizontal acceleration")
plt.title("Horizontal movement intensity")
plt.grid(True)
plt.show()
```

```python
threshold = 0.4

df["is_moving"] = df["acc_horizontal"] > threshold

plt.figure(figsize=(12, 5))
plt.plot(df["time_insert"], df["acc_horizontal"], label="acc_horizontal")
plt.axhline(threshold, linestyle="--", label="threshold")

plt.xlabel("time (s)")
plt.ylabel("horizontal acceleration")
plt.title("Horizontal movement detection")
plt.legend()
plt.grid(True)
plt.show()
```
