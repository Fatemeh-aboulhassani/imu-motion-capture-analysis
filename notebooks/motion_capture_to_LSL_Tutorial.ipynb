# IMU and Motion Capture Recording with LSL

In this project, I wanted to record data from IMU sensors and a motion capture system at the same time.  
For this, I used **Lab Streaming Layer (LSL)**, because it allows different systems to send their data as streams and then record everything together with **LabRecorder**.

The general idea is:

```text
IMU sensors  →  LSL stream
Qualisys/QTM →  LSL stream

Both streams → LabRecorder → one XDF file

Software and tools

For this setup I used:

Anaconda / Conda
Python
LabRecorder
Qualisys Track Manager (QTM)
Qualisys LSL App
Python packages like pylsl and qtm-rt

The IMU data is streamed with a Python script.
The Qualisys data is streamed with the Qualisys LSL App.

Setting up the Qualisys LSL environment

First, I created a separate conda environment for the Qualisys LSL app:

* conda create -n qualisys_lsl python=3.10

Then I activated the environment:

conda activate qualisys_lsl

After that, I installed the required Python packages:

* python -m pip install qtm-rt pylsl

Here, qtm-rt is needed to communicate with QTM, and pylsl is needed for working with LSL streams.

Installing the Qualisys LSL App

I cloned the official Qualisys LSL App from GitHub:

* git clone https://github.com/qualisys/qualisys_lsl_app.git

Then I went into the folder:

cd qualisys_lsl_app

and installed the requirements:

* python -m pip install -r requirements.txt

To start the graphical user interface, I used:

* python -m qlsl.gui

This opens the Qualisys LSL App, where the connection to QTM can be configured.

Starting the setup again

After everything is installed, I do not need to repeat the installation steps every time.

To start the Qualisys LSL App again, I only use:

* conda activate qualisys_lsl
* cd "C:\Users\abolhassni\Desktop\TU Darmstadt\Term3\ANSYMB\LSL\qualisys_lsl_app"
* python -m qlsl.gui

The path has to be changed depending on where the project is saved.
