```python

# Static calibration result: gravity direction from standing
gravity_vectors = {
    "Pelv":   np.array([-0.9822346,   0.00794906,  0.18748905]),
    "lLeg_L": np.array([-0.9977788,  -0.0434721,  -0.05047445]),
    "uLeg_L": np.array([-0.97699046,  0.00125464, -0.21327917]),
    "Foot_L": np.array([-0.2089611,   0.05583238,  0.97632885]),
}

calibration_axes = {}

for sensor_name in functional_axis.keys():

    # 1) Vertical axis from static calibration
    v = gravity_vectors[sensor_name]
    v = v / np.linalg.norm(v)

    # 2) Functional axis from squat PCA
    f = functional_axis[sensor_name]
    f = f / np.linalg.norm(f)

    # 3) Remove the component of the functional axis along the vertical axis
    f_perp = f - np.dot(f, v) * v
    f_perp = f_perp / np.linalg.norm(f_perp)

    # 4) Compute the third orthogonal axis
    third = np.cross(v, f_perp)
    third = third / np.linalg.norm(third)

    # 5) Recompute the functional axis to ensure orthogonality
    f_corrected = np.cross(third, v)
    f_corrected = f_corrected / np.linalg.norm(f_corrected)

    # 6) Build calibration matrix
    R_calib = np.column_stack((v, f_corrected, third))

    calibration_axes[sensor_name] = {
        "vertical": v,
        "functional": f_corrected,
        "third": third,
        "R_calib": R_calib
    }

    print(sensor_name)
    print("vertical axis:")
    print(v)
    print("functional axis corrected:")
    print(f_corrected)
    print("third axis:")
    print(third)
    print("calibration matrix:")
    print(R_calib)
    print()
```
